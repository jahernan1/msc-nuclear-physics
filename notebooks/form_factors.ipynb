```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
ffargold=np.loadtxt('../results/densities/Ar40_fsugold_WeakFF.dat')
ffargarnet=np.loadtxt('../results/densities/Ar40_fsugarnet_WeakFF.dat')
ffarr022=np.loadtxt('../results/densities/Ar40_rmf022_WeakFF.dat')
ffarr028=np.loadtxt('../results/densities/Ar40_rmf028_WeakFF.dat')
ffarr032=np.loadtxt('../results/densities/Ar40_rmf032_WeakFF.dat')
ffarta=np.loadtxt('../results/densities/Ar40_tamufsua_WeakFF.dat')
ffartc=np.loadtxt('../results/densities/Ar40_tamufsuc_WeakFF.dat')
```

```python
arskins = np.loadtxt('../results/skins/Ar40_neutron_skins.txt', usecols=2)
print(arskins)
armin = np.argmin(ffargold[:,1])
```

```python
plt.plot(ffargold[:,0], ffargold[:,1], label='fsugold({:.3f})'.format(arskins[0]))
plt.plot(ffargarnet[:,0], ffargarnet[:,1], label='fsugarnet({:.3f})'.format(arskins[1]))
plt.plot(ffarr022[:,0], ffarr022[:,1], label='rmf022({:.3f})'.format(arskins[2]))
plt.plot(ffarr028[:,0], ffarr028[:,1], label='rmf028({:.3f})'.format(arskins[3]))
plt.plot(ffarr032[:,0], ffarr032[:,1], label='rmf032({:.3f})'.format(arskins[4]))
plt.plot(ffarta[:,0], ffarta[:,1], label='tamufsua({:.3f})'.format(arskins[5]))
plt.plot(ffartc[:,0], ffartc[:,1], label='tamufsuc({:.3f})'.format(arskins[6]))
plt.legend()
plt.xlim(0, ffargold[:,0][armin])
plt.suptitle(r'$^{40}Ar$')
plt.xlabel('$q\ [fm^{-1}]$')
plt.ylabel('$F_{weak}$')
plt.savefig('Ar40_WeakFF.png')
```

```python
plt.plot(ffargold[:,0], ffargold[:,2]-ffargold[:,1], label='fsugold({:.3f})'.format(arskins[0]))
plt.plot(ffargarnet[:,0], ffargarnet[:,2]-ffargarnet[:,1], label='fsugarnet({:.3f})'.format(arskins[1]))
plt.plot(ffarr022[:,0], ffarr022[:,2]- ffarr022[:,1], label='rmf022({:.3f})'.format(arskins[2]))
plt.plot(ffarr028[:,0], ffarr028[:,2]- ffarr028[:,1], label='rmf028({:.3f})'.format(arskins[3]))
plt.plot(ffarr032[:,0], ffarr032[:,2]- ffarr032[:,1], label='rmf032({:.3f})'.format(arskins[4]))
plt.plot(ffarta[:,0], ffarta[:,2]- ffarta[:,1], label='tamufsua({:.3f})'.format(arskins[5]))
plt.plot(ffartc[:,0], ffartc[:,2]-ffartc[:,1], label='tamufsuc({:.3f})'.format(arskins[6]))
plt.legend()
plt.yscale('symlog')
plt.xlim(0, 3)
plt.suptitle(r'$^{40}Ar$')
plt.xlabel('$q\ [fm^{-1}]$')
plt.ylabel('$F_{charge}-F_{weak}$')
plt.savefig('Ar40_FFdiff.png')
```
